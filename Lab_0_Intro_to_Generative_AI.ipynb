# Lab 0: Intro to Generative AI

In this notebook you will make a single API call to a LLM, print the response, and display the token usage.

> **Note**: Replace `YOUR_API_KEY` with a valid OpenAI (or compatible) API key before running.

```python
import os
import openai

# Set your API key – keep it secret!
openai.api_key = os.getenv("OPENAI_API_KEY", "YOUR_API_KEY")

prompt = "Write a short Python function that returns the square of a number."

response = openai.ChatCompletion.create(
    model="gpt-3.5-turbo",
    messages=[{"role": "user", "content": prompt}],
    temperature=0.2,
    max_tokens=150,
    n=1,
)

answer = response["choices"][0]["message"]["content"]
usage = response["usage"]

print("--- Model answer ---\n")
print(answer)
print("\n--- Token usage ---")
print(f"Prompt tokens: {usage['prompt_tokens']}")
print(f"Completion tokens: {usage['completion_tokens']}")
print(f"Total tokens: {usage['total_tokens']}")
```
