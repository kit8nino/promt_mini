# Lab 3: Task Decomposition and Prompt Chaining

This lab demonstrates how to split a complex request into atomic steps and chain the results.


```python
import os, json
import openai

openai.api_key = os.getenv("OPENAI_API_KEY", "YOUR_API_KEY")

# Helper to call the model
def ask(prompt):
        resp = openai.ChatCompletion.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": prompt}],
            temperature=0,
            max_tokens=300,
        )
        return resp["choices"][0]["message"]["content"].strip()
```

### Step 1 – Read CSV and return list of records

We ask the model to read a CSV (provided as a string) and output a JSON array of objects.


```python
csv_data = """
name,age
Alice,30
Bob,22
Charlie,17
"""

step1_prompt = (
    "Ты — senior Python‑разработчик.\n"
    "Контекст: CSV‑строка ниже.\n"
    "Задача: прочитать CSV и вернуть список словарей {name, age}.\n"
    "Формат вывода: JSON‑массив.\n"
    f"CSV:\n{csv_data}\n"
)

step1_result = ask(step1_prompt)
print(step1_result)
```

### Step 2 – Filter records where `age` > 18

We feed the JSON from step 1 back into the model.


```python
step2_prompt = (
    "Ты — senior Python‑разработчик.\n"
    "Контекст: результат шага 1 ниже.\n"
    "Задача: отфильтровать элементы, где `age` > 18.\n"
    "Формат вывода: JSON‑массив.\n"
    f"Input JSON:\n{step1_result}\n"
)

step2_result = ask(step2_prompt)
print(step2_result)
```

### Step 3 – Return the filtered data as a pretty‑printed JSON string


```python
step3_prompt = (
    "Ты — senior Python‑разработчик.\n"
    "Контекст: результат шага 2 ниже.\n"
    "Задача: вывести его как строку JSON без дополнительных пояснений.\n"
    "Формат вывода: строка JSON.\n"
    f"Input JSON:\n{step2_result}\n"
)

final_answer = ask(step3_prompt)
print(final_answer)
```
