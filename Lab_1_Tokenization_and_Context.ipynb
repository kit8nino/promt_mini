# Lab 1: Tokenization and Context Window

In this lab you will:
1. Tokenize a long paragraph using `tiktoken` (or the tokenizer of your LLM).
2. Observe how many tokens exceed a typical 4 096‑token context window.
3. Visualise a sliding‑window view of the token sequence.


```python
import tiktoken
from textwrap import dedent

# Example long text – repeat a sentence to exceed the window
base_sentence = "Artificial intelligence enables new possibilities in many domains. "
long_text = base_sentence * 500  # ~500 * 12 ≈ 6000 tokens (depends on tokenizer)

enc = tiktoken.get_encoding("cl100k_base")  # tokenizer used by gpt‑3.5‑turbo
tokens = enc.encode(long_text)
print(f"Total tokens in the paragraph: {len(tokens)}")

# Show first 20 tokens for illustration
print("First 20 tokens:", tokens[:20])

# Truncate to a 4,096‑token context window
window_size = 4096
truncated = tokens[:window_size]
print(f"Tokens after truncation to {window_size}: {len(truncated)}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a rolling‑window count (how many tokens are visible at each step)
step = 200  # slide by 200 tokens each frame
positions = list(range(0, len(tokens) - window_size + 1, step))
visible_counts = [window_size for _ in positions]

plt.figure(figsize=(8, 2))
plt.plot(positions, visible_counts, drawstyle='steps-post')
plt.title('Sliding Context Window (size = 4096 tokens)')
plt.xlabel('Start token index')
plt.ylabel('Visible tokens')
plt.ylim(0, window_size + 500)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
