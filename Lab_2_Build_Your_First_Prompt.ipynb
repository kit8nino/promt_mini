# Lab 2: Build Your First Prompt

In this notebook you will construct a prompt using the three‑component template, add a few‑shot example, and request a **step‑by‑step** answer from the model.


```python
import os, json
import openai

openai.api_key = os.getenv("OPENAI_API_KEY", "YOUR_API_KEY")

# System prompt – role definition (short, ≤30 tokens)
system_prompt = (
        "Ты — senior backend‑разработчик. Отвечай в формате JSON без лишних слов."
    )

# Few‑shot example (Input → Output)
few_shot = (
        "Input: Напиши функцию, которая принимает список чисел и возвращает их среднее.\n"
        "Output: {\n"
        "  \"function_name\": \"mean\",\n"
        "  \"code\": \"def mean(nums): return sum(nums) / len(nums)\",\n"
        "  \"description\": \"Вычисляет среднее арифметическое\"\n"
        "}\n"
    )

# User request
user_prompt = (
        "Input: Напиши функцию, которая принимает список строк и возвращает их в алфавитном порядке.\n"
        "think step‑by‑step, then provide the final JSON."
    )

full_prompt = f"System: {system_prompt}\n\n{few_shot}\n{user_prompt}"

response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=[{"role": "user", "content": full_prompt}],
        temperature=0,
        max_tokens=300,
    )

answer = response["choices"][0]["message"]["content"]
print(answer)
```

## What to observe

- The model should output **pure JSON** (no surrounding text).
- The reasoning steps are omitted from the final answer because we asked for `think step‑by‑step` before the JSON.
- Try modifying the system prompt to add a length limit or a different persona and re‑run.
